```
클래스 메서드

def function(cls):

클래스 자체의 데이터를 관리할 때


정적 메서드
@staticmethod

def function():

객체 상태와 관계없는 독립적인 기능을 만들 때
```

````python
# 상품에 대한 클래스
class Product:
    # 클래스의 변수: 다음에 줄 번호표?
    _next_id = 1

    def __init__(self, name, price, stock):
        # 번호표 받기
        self.id = Product._next_id
        # 다음 사람을 위해 번호표
        Product._next_id += 1
        self.name = name
        self.price = price
        # 직접 수정 금지의 표시 _언더바
        self._stock = stock

    # 재고확인 재고읽기?
    def get_stock(self):
        return self._stock
    # 품절인지 아닌지 확인
    def is_sold_out(self):
        return self._stock == 0
    # 재고 줄이기 유일한 통로
    def reduce_stock(self, pty):
        self._stock -= pty

    def __str__(self):
        if self.is_sold_out():
            stock_text = "품절"
        else:
            stock_text = f"재고 {self._stock}개"
        return f"{self.id}. {self.name:<10} {self.price:>8}원 ({stock_text})"

# -------------------------- 테스트 코드 ---------------------------
a = Product("무선 이어폰", 89000, 0)
b = Product("텀블러", 15000, 10)
print(a)    # 1. 무선 이어폰 ...
print(b)    # 2. 텀블러

# 할인 쿠폰의 대한 클래스
class Discount:
    # 부모 클래스 쿠폰 없음
    def __init__(self, name="쿠폰 없음"):
        self.name = name

    # total = 상품 합계 return = 0 항상 0원
    def discount_amount(self, total):
        return 0

# 자식 클래스 비율 할인
class RateDiscount(Discount):
    def __init__(self, rate, max_count):
        super().__init__(f"{round(rate*100)}% 할인 (최대 {max_count:,}원)")
        self.rate = rate
        self.max_count = max_count

    # 오버라이딩    
    def discount_amount(self, total):
        # 원 단위 미만 버림 => int()
        amount = int(total * self.rate)
        # 최대 금액을 넘지 않게 설정
        return min(amount, self.max_count)


class FixedDiscount(Discount):
    def __init__(self, amount, min_total):
        super().__init__(f"{amount:,}원 할인 ({min_total:,}원 이상 구매 시)")
        self.amount = amount
        self.min_total = min_total

    # 오버라이딩
    def discount_amount(self, total):
        if total >= self.min_total:
            return self.amount
        return 0

coupons = [Discount(), RateDiscount(0.1, 5000), FixedDiscount(3000, 20000)]
for c in coupons:
    print(c.name, "→", c.discount_amount(313000))
# 쿠폰 없음 → 0
# 10% 할인 (최대 5,000원) → 5000   (31,300원이지만 최대 5,000원)
# 3,000원 할인 (20,000원 이상 구매 시) → 3000
````
