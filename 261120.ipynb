```python
amounts = [3200, 1500, 15000, 6500, 32000, 4800,
           10900, 1500, 2700, 3000, 18500, 7000]


print(amounts)
print("합계" , sum(amounts))

print(amounts * 2)
print("==="*3)
```

```python
import numpy as np

arr = np.array(amounts)         # numpy로 생성된 array 함수
arr * 2

print("합계:", arr.sum())
print("평균;", round(arr.mean()))
print("최소;", arr.min())
print("최대;", arr.max())

print("모양:",arr.shape, "데이터 타입:", arr.dtype)
```

```python
print(arr >= 10000)
arr[arr >= 10000]           # 조건을 만족하는 경우만 출력
```

```python
amounts = [3200, 1500, 15000, 6500, 32000, 4800,
           10900, 1500, 2700, 3000, 18500, 7000]

shops = ["편의점", "버스", "영화관", "분식집", "온라인쇼핑", "카페",
         "음원앱", "버스", "편의점", "웹툰", "온라인쇼핑", "분식집"]

categories = ["식비", "교통", "문화", "식비", "쇼핑", "식비",
              "문화", "교통", "식비", "문화", "쇼핑", "식비"]

print(shops)
food = 0
for c, a in zip(categories, amounts):
    if c == "식비":
        food = food + a
print(food)
```

```python
import pandas as pd

spend = pd.DataFrame({
    "day": [2, 3, 4, 5, 7, 9, 12, 14, 16, 19, 20, 25],
    "shop": shops,
    "category": categories,
    "amount": amounts
})
spend.head(5)           # 상위 5개 추출
# spend.info()
# print(spend.shape)
```

```python
spend["amount"]
print("합계:", spend["amount"].sum())
print("평균:", round(spend["amount"].mean(),2))
```

```python
# spend["shop", "amounts"] "shop", "amounts" 이렇게 한 컬럼으로 인식함 이렇게 하면 안됨
# spend[ 컬럼명 ]
# spend[ [컬럼1, 컬럼2, ...] ]
spend[ ["shop", "amount"] ]         # pandas에서는 이렇게 데이터 뽑아옴
```

### 조건 설정

```python
# 조건식이 됨 아래 문장이
spend["category"] == "문화"
```

```python
# spend[ 조건 ] 조건이 True인 행만 return
# spend[ "컬럼명" ] -> "컬럼1"의 로우(row) 모두 리턴
spend[spend["category"] == "문화"]
```

```python
# 조건식
# spend[ 조건식 ]
# spend["amount"] >= 10_000
spend[spend["amount"] >= 10_000]
```

```python
# 식비 5000원 이상 사용한 행을 출력
spend[(spend["category"] == "식비") & (spend["amount"] >= 5000)]
```

```python
top3 = spend.sort_values("amount", ascending=False).head(3)
top3
```

```python
# spend["새 컬럼"] = 계산식      이렇게 새로운 컬럼 추가 가능
spend["point"] = spend["amount"] // 100 # 100원 마다 1 포인트        3200 나누기 32  :   3200//100   (몫) 3200 % 100 (나머지)
spend.head(5)
print(spend["point"].sum())

spend
```
