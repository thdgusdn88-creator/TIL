````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt    # 통계 그래프 라이브러리
import seaborn as sns


plt.rcParams["font.family"] = "Malgun Gothic"

df = sns.load_dataset("diamonds")
print(df.shape)
df.info()
# df.describe()
````

````python
df.describe()

# 평균이 더 큼 이 차이는 평균값을 유의미하게 만들지 못하게 하는 이상치 값들이 있을 수 있다는 것임
````

````python
zero_mask = (df["x"] == 0) | (df["y"] == 0) | (df["z"] == 0)
print(f"\n[2] 셋 중 하나라도 0인 행 수: {zero_mask.sum()}")
df[zero_mask]
````

````python
big_mask = (df["y"] > 20) | (df["z"] > 20)
print(f"\n[3] y 또는 z가 20mm 초과인 행 수: {big_mask.sum()}")
df[big_mask]
````

````python
clean = df[~(zero_mask | big_mask)].copy()
print(f"\n[4] 제거 전 행 수: {len(df)}")
print(f"    제거 후 행 수: {len(clean)}")
print(f"    제거된 행 수: {len(df) - len(clean)}")

# 머신러닝에 방해 될 수 있는 빈칸을 지웠음
# 실제로 비싼 다이아몬드는 존재해서
````

````python
price_mean = clean["price"].mean()
price_median = clean["price"].median()

fig, ax = plt.subplots(figsize=(9, 5))
sns.histplot(data=clean, x="price", bins=50, ax=ax)
ax.axvline(price_mean, color="red", linestyle="--", label=f"평균 {price_mean:,.0f}")
ax.axvline(price_median, color="green", linestyle="-", label=f"중앙값 {price_median:,.0f}")
ax.set_title("price 분포")
ax.set_xlabel("가격 (USD)")
ax.legend()
plt.show()

print(f"평균: {price_mean:,.1f} / 중앙값: {price_median:,.1f}")
````

````python
sample = clean.sample(3000, random_state=42)

fig, ax = plt.subplots(figsize=(8, 5))
sns.scatterplot(data=sample, x="carat", y="price", alpha=0.5, ax=ax)
ax.set_title("carat vs price (무작위 3,000개)")
plt.show()
````

````python
corr = clean.select_dtypes("number").corr()
price_corr = corr["price"].sort_values(ascending=False)
print(price_corr)

# price와 상관계수가 숫자열 중 가장 높아서
# carat이 커질 수 록 퍼져서 직선 맞추기 어려움
# 그리고 0.5 마다 새로로 분포해서 직선 사용 어려움
````

````python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(clean[["carat"]], clean["price"])

w = model.coef_[0]
b = model.intercept_
print(f"w (기울기): {w:,.1f}")
print(f"b (절편)  : {b:,.1f}")
````

### w는 1케럿당 오르는 가격의 수치정도
### 데이터 값에 따라 정확도는 나뉨

````python
new_data = pd.DataFrame({"carat": [0.25, 0.5, 1.0, 2.0]})
pred = model.predict(new_data)

for c, p in zip(new_data["carat"], pred):
    print(f"{c:>4} 캐럿 → 예측 가격 {p:>10,.0f} 달러")


# 그래프는 계속 일직선인데 실제 가격은 작은 캐럿에서 완만하게 휘어서 0 근처로 모이는 휘어짐을 못 따라가서 절편(b)가 음수가 돼서
````

````python
sample = clean.sample(3000, random_state=42)

line_x = pd.DataFrame({"carat": [0.2, 3.0]})
line_y = model.predict(line_x)

fig, ax = plt.subplots(figsize=(9, 6))
sns.scatterplot(data=sample, x="carat", y="price", alpha=0.4, ax=ax)
ax.plot(line_x["carat"], line_y, color="red", linewidth=2, label="회귀 직선")
ax.axhline(clean["price"].mean(), color="gray", linestyle="--", label="평균 가격")
ax.set_title("carat → price 선형회귀")
ax.set_xlabel("carat")
ax.set_ylabel("price (USD)")
ax.legend()
plt.show()
````

````python
from sklearn.linear_model import LinearRegression

num_cols = ["carat", "depth", "table", "x", "y", "z"]
sample = clean.sample(3000, random_state=42)

fig, axes = plt.subplots(2, 3, figsize=(16, 9))

rows = []

for ax, col in zip(axes.ravel(), num_cols):
    # 1) 이 열 하나로 직선 학습
    model = LinearRegression().fit(clean[[col]], clean["price"])
    w = model.coef_[0]
    b = model.intercept_

    # 2) 직선의 양 끝 두 점 계산
    x_min, x_max = clean[col].min(), clean[col].max()
    line_x = pd.DataFrame({col: [x_min, x_max]})
    line_y = model.predict(line_x)

    # 3) 산점도 + 직선 그리기
    sns.scatterplot(data=sample, x=col, y="price", alpha=0.3, s=10, ax=ax)
    ax.plot(line_x[col], line_y, color="red", linewidth=2)
    ax.set_title(f"{col} → price (w={w:,.0f})")

    # 4) 표에 넣을 한 줄 기록
    rows.append({
        "열": col,
        "w(기울기)": round(w, 1),
        "최솟값": x_min,
        "최댓값": x_max,
        "범위(최대-최소)": round(x_max - x_min, 2),
    })

plt.tight_layout()
plt.show()

# ─────────────────────────────────────
# 표 하나로 정리
# ─────────────────────────────────────
summary = pd.DataFrame(rows)
print(summary.to_string(index=False))

# ─────────────────────────────────────
# M3 상관계수와 비교
# ─────────────────────────────────────
corr_price = clean[num_cols + ["price"]].corr()["price"].drop("price")
print("\nprice와의 상관계수 (큰 순서)")
print(corr_price.sort_values(ascending=False))
````

```
열을 비교할 때는 w 대신

① 상관계수 (가장 간단한 방법)
-1~1 사이의 값으로, 단위와 범위에 영향을 받지 않음
절댓값이 1에 가까울수록 직선 관계가 강함 
z는 약 0.88, table은 약 0.13이라 z가 훨씬 강한 관계임을 바로 알 수 있음

② R² (결정계수)
"이 열 하나로 가격 변화의 몇 %를 설명하는가"를 나타냄 
단일 열 직선에서는 상관계수의 제곱과 같음 
(예: 0.88² ≈ 0.77이면 가격 변화의 약 77%를 설명, 0.13² ≈ 0.02이면 약 2%)

③ 산점도 그림
점이 직선 주변에 모여 있는지 눈으로 확인
곡선, 퍼짐 같은 숫자로 안 보이는 문제도 그림에서만 보임

④ 표준화한 기울기 (선택)
열을 같은 단위(평균 0, 표준편차 1)로 맞춘 뒤 w를 구하면 열끼리 공정하게 비교
```
