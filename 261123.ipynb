````python
import pandas as pd
import numpy as np
import seaborn as sns
# 확장 프로그램이 잘 설치 됐는지 버전 확인

print("pandas", pd.__version__, "numpy", np.__version__, "seaborn", sns.__version__)
````

````python
# seaborn에 들어가 있는 타이타닉 데이터를 df 데이터 프래임 변수로 불러오기
df = sns.load_dataset('titanic')
df.head(5)
````

````python
# 표의 크기 확인
df.shape
````

````python
# 칸마다 값이 몇개 있는지, 자료형이 무엇인지 보기
df.info()
````

````python
# 셀 중에 숫자 칸의 평균, 사분위수, 최댓값 보기
df.describe()
````

````python
# 셀의 빈칸 개수 확인
df.isnull().sum()
````

````python
# 나이가 비어 있는 승객 몇명인지 확인하고 어떤사람인지 확인

empty_age = df["age"].isnull()
print("나이 빈칸:", empty_age.sum(), "명")
df.loc[empty_age, ["sex", "age", "class"]].head()
````

````python
# 평균과 중앙값
print("평균:", round(df["age"].mean(), 2), "| 중앙값:", df["age"].median())


mean_age = df["age"].mean()

# 평균값으로 셀 채우기
df.loc[:, "age"] = df["age"].fillna(mean_age)
print("남은 빈칸:", df["age"].isnull().sum())

# 승객드 나이 바뀐거 확인
df.loc[empty_age, ["sex", "age", "class"]].head()
````

````python
# 탑승항구는 글자라 평균이 없습니다. 가장 많이 탄 항구로 채우기
# 항구별 승객 수를 빈칸 까지 포함해서 세보기

df["embark_town"].value_counts(dropna=False)
````

````python
# 최빈값( 가장 많은 빈도로 사용된 것)

mode_town = df["embark_town"].mode()[0]
print("최빈값:", mode_town)
df.loc[:, "embark_town"] = df["embark_town"].fillna(mode_town)
# 다시 채워졌는지 확인

df["embark_town"].value_counts(dropna=False)
````

````python
# 앞글자 이니셜 따오기

df.loc[:, "embarked"] = df["embark_town"].str[0]
df[["embarked", "embark_town"]].isnull().sum()
````

````python
# dask 갑판 빈칸을 앞 뒤 승객 값으로 채운 사본 만들기

test = df["deck"].ffill().bfill()
pd.DataFrame({"원본": df["deck"].value_counts(), "ffill·bfill 후": test.value_counts()})

# .ffill()
# pandas
# forward fill. 빈칸을 바로 위 행의 값으로 채웁니다. 시간 순서가 있는 데이터(어제 값으로 오늘 빈칸)에 쓰는 방법입니다.
# .bfill()
# pandas
# backward fill. 빈칸을 바로 아래 행의 값으로 채웁니다. ffill 로 못 채운 맨 앞 빈칸을 마저 채울 때 함께 씁니다.
````

````python
# deck의 자료형과 받을 수 있는 값의 목록 범주를 확인

print(df["deck"].dtype)
df["deck"].cat.categories
````

````python
# deck에 Unkonwn 범주를 추가하고 그것을 빈칸으로 채운 뒤, 범주 목록 확인

df["deck"] = df["deck"].cat.add_categories("Unknown").fillna("Unknown")
print(df["deck"].cat.categories.tolist())
print(df["deck"].value_counts())
df.isnull().sum().sum()
````

````python
# 요금의 Q1 · Q3 · IQR 을 구하고, Q3 + 1.5×IQR 보다 큰 승객이 몇 명인지 세세요.\

print(df["fare"].quantile([0.25, 0.5, 0.75]))
df["fare"].describe()[["25%", "50%", "75%"]]
````

````python
# Q1, Q3, IQR 위쪽 한계를 계산하고, 한계를 넘는 승객 수와 등급 확인

q1 = df["fare"].quantile(0.25)
q3 = df["fare"].quantile(0.75)
iqr = q3 - q1
upper = q3 + 1.5 * iqr
print(f"Q1={q1:.2f}, Q3={q3:.2f}, IQR={iqr:.2f}, 위쪽 한계={upper:.2f}")
print("이상치 후보:", (df["fare"] > upper).sum(), "명")
df.loc[df["fare"] > upper, "class"].value_counts()
````

````python
# 300달러 이상만 거르고 행 수 확인

before = len(df)
df = df[df["fare"] < 300].copy()
print(before, "→", len(df), "명")
````

````python
# 완전히 같은 행이 몇 개인지 세고, 몇개를 눈으로 보기

print("중복 행:", df.duplicated().sum(), "개")
df[df.duplicated(keep=False)].sort_values(["fare", "age"]).head(6)
````
