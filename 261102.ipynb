##### git, github, gh 서로 다른 도구임
- gh는 github를 터미널에서 사용하는 도구
- winget install --id GitHub.cli -e -> 먼저 코드 가져와서 설치 github.cli 홈페이지 가서
```
~ $ gh auth login
? Where do you use GitHub? GitHub.com
? What is your preferred protocol for Git operations on this host? HTTPS
? Authenticate Git with your GitHub credentials? Yes
? How would you like to authenticate GitHub CLI?  [Use arrows to move, type to filter]
> Login with a web browser
  Paste an authentication token
```
- 브라우저 가서 직접 로그인하기
- gh browse -> 이 명령어로 해당 연결된 원격 repository 브라우저로 열 수 있음
- gh repo list -> 내 계정의 repo 목록을 보고 clone나 delete 하려고 할 때 이름 복사하기 가능
