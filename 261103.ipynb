```python
import random                # import를 사용해 난수 발생시키는 모듈을 가져옴

number = random.randint(1, 100)     # 1~ 100 사이의 난수를 number에 지정

try_count = 0                       # 시도 횟수

print("="*45)
print("컴퓨터가 1부터 100까지의 숫자를 생성했습니다.")   # 타이틀 화면 띄워주기
print("="*45)


while True:                 # 메인 루프
    pass    # pass는 지금은 비워두지만 추후 쓰려고 비워둔 자리 ex) 추가 구문 작성 예정, TODO
    user_input = input("컴퓨터의 숫자를 맞춰보세요 (1 ~ 100):") # input은 문자열 입력값을 받음 그래서 int로 받은 값을 정수로 바꾸기

    if user_input == "q":                               # 문자열 q를 입력받으면 동작중지
        print(f"게임을 종료합니다. 정답은 {number}입니다.")  # f를 넣으면 문자열 안에 변수나
        break                                            # 표현식을 넣을 수 있음 {} 사용

    if not user_input.isdecimal():   #변수 값이 정수 값이 아니면 False 출력 정수면 True 
        print("숫자만 입력해주세요.")   # not을 사용해서 숫자가 아니면 print 출력
        continue

    user_number = int(user_input)   #input으로 받은 문자자를 정수로 바꿈

    if user_number <= 0 or user_number > 100:   #100이상 또는 0이하의 값을 차단
        print("1~100 사이값을 입력하세요.")
        continue

    try_count += 1  # 시도 횟수 증가 누적치 이 코드는 이 아래 있는 코드보다만 위에 있으면 됨

    if user_number < number:            # 입력값과 정답을 비교    
        print("up! 더 큰 수 입니다")
    elif user_number > number:
        print("down! 더 작은 수 입니다")
    else:                             # else는 바로 뒤에 클론
        print(f"정답압나다. {try_count}회 만에 성공하셨습니다.")
        break
```

```python
# 회원 데이터
user_database = [
    {"name": "손흥민", "level": "VIP", "active": True, "point": 15000},
    {"name": "김민재", "level": "Gold", "active": False, "point": 9000},
    {"name": "조규성", "level": "VIP", "active": True, "point": 12000},
    {"name": "이강인", "level": "Gold", "active": True, "point": 8000},
    {"name": "홍명보", "level": "Broze", "active": False, "point": 1000}
]

# VIP 포인트 누적 변수
total_vip_points = 0
# 현재 활성 회원 수
active_user_count = 0

print("=== 마케팅 대상 활성 VIP 및 Gold 회원 결과")

for user in user_database:              # for 변수 in 반복할 데이터:
    # 활성화 된 고객 필터링 active
    if user["active"]:     # 불리언 타입 구별 / keys나 values 값을 가져올때는 [] 사용
        # 활성 회원수 누적
        active_user_count += 1

        if user["level"] == "VIP":
            total_vip_points += user["point"]

            print(list(f"[VIP] {user['name']} 회원님에게 VIP 쿠폰이 발행되었습니다.")) # list로 변환 후 출력
        elif user["level"] == "Gold":
            print(f"[Gold] {user['name']} 회원님에게 Gold 쿠폰이 발행되었습니다.")  # ctrl + d 멀티커서


print(f"전체 활성 회원 수 : {active_user_count}명")
print(f"활성 VIP 포인트 누적 : {total_vip_points}점")
```

#### 딕셔너리 주요 메서드

- keys(), values(), item() -> 리스트 형태로 보임

```python
menu = {"coffee" : 4000, "tea": 4500}

# key 목록 조회
print(menu.keys())
# values 목록 조회
print(menu.values())
# item 조회 (키, 값) 튜플 형태로 리턴
print(menu.items())

# 딕셔너리는 인덱서 방식으로 값을 참조 불가 -> list로 변환
print(list(menu.keys())[1])

print(sum(menu.values()))
```

```python
print("coffee" in menu)        # True
print(4000 in menu)            # Fulse -> dicionary에서 in은 key를 검사함

print( 4000 in menu.values())  # True

# 기본적으로 다른 쉘이지만 위에서 변수 할당? 지정을 해주면 아래까지도 영향이 감
```

```python
# menu["cake"]
price = menu.get("cake", 5000)

print(price)                # 케이크는 없으니까 두번째 인자값 5000반영

# 값이 없는 경우 None
print(menu.get("cake"))     # 케이크가 없으니 none 출력

print(menu.get("coffee",0))   #커피는 실제로 있는 값이라 기본값 0은 무시되고 실제 값 반영

print(menu)
```

#### 집합 (set)

수학의 집합 : 중복 없고, 순서 없는 자료형 -> 인덱스로 꺼낼 수 없음 [0] 이런식으로 불가능

- 중괄호 사용
- set() 함수를 생성

```python
grade = set([1, 2, 2, 3, 3, 3])

print(grade)
```

```python
print(type({})) # {} 빈 딕셔너리

print(type(set())) # set() 빈 집합 자료형

print({1, 2, 2}) # 집합 -> 중괄호 안에 값만 나열되어 있으면 집합
```

```python
s = {1, 2}

s.add(2)
s.add(5)

print(s)

s.remove(1) # 없는 키 삭제하면 오류가 나옴
print(s)

s.discard(2) # 없는 키 삭제해도 오류발생 X
print(s)
```

```python
set_a = {1, 2, 3}
set_b = {3, 4, 5}

print(set_a & set_b) #교집합   & 엔퍼센트 _ 언더스코어 * 에스터리스크
print(set_a | set_b) #합집합    | 파이프
print(set_a - set_b) #차집합
print(3 in set_a) #포함 여부
```
