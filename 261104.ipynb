```python
# 평균 계산해서 반환하는 함수
# 평균 = 총합 / 갯수

def get_average(*scores):           # 가변인수 사용
    if len(scores) == 0:            # len() 객체의 길이를 정수로 돌려줌
        return 0                    # 이때 안에 값이 없으면 함수 종료

    # if not scores:
    #     return
    
    # 합계 계산
    total = sum(scores)
    # 평균 계산
    avg = total / len(scores)
    # 결과값 반환
    return avg

result1 = get_average(80,90,100)
result2 = get_average(75, 80, 92, 100, 55)
result3 = get_average()

print(f"3과목 평균: {result1}")
print(f"5과목 평균: {result2}")
```

```python
# 전역 변수
value = "전역 데이터"                       # 1


# function============================
def outer_function():
    value = "지역 변수 1"                   # 3 지역변수 1 할당


    def inner_function():                  
        value = "지역 변수 2"               # 5 지역변수 2 할당
        print(value)                       # 6 지역변수 2 프린트

    inner_function()                       # 4 inner 호출
    print(value)                           # 7 지역변수 1 프린트
# function=============================

outer_function()                            # 2 outer 호출
print(value)                            # 8 전역변수 전역데이터 프린
```

```python
# global 키워드

counter = 0             # 전역 변수

def inc_counter():
    # counter = 0
    global counter      # 전역 변수를 참조한다 global 명령어를 통해서
    counter += 1


inc_counter()
inc_counter()

print(counter)
```

```python
# map 모든 요소를 반환
number = [1, 2, 3, 4, 5]

result = list(map(lambda x: x + 10, number))   # map(,number)을 통해서 x에 가져옴
print(f"계산 결과: {result}")             # 리스트 형식으로 안하면 출력이 안됨
                                        # 애초에 number이 리스트라
# 리스트 컴프리핸션
print([x+10 for x in number])

filter_result = list(filter(lambda x: x % 2 == 0, number))  # filter(,number) True, Fulse 값일 경우
print(f"짝수 결과: {filter_result}")

print([x for x in number if x % 2 == 0])
```

```python
players = [
    ("손흥민", 95),
    ("김민재", 60),
    ("이강인", 80)
]

print(sorted(players, key=lambda p: p[1], reverse=True))      #내림차순
print(sorted(players, key=lambda p: p[1]))                    #오름차순
print(max(players, key=lambda p: p[1]))                       #최고점수 호출
```
